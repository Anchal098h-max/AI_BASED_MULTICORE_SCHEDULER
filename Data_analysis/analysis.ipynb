```python
fig , axes = plt.subplots( 4 ,4 ,   figsize = ( 20 , 12))
axes = axes.flatten()

queue_len = [
    col   for col in df.columns 
     if  "queue_length" in col.lower()
]
for i ,  col in enumerate(queue_len) : 
    axes[i].plot(pd.to_datetime(df['timestamp']), df[col], linewidth=1)
    axes[i].set_title( col)
    axes[i].set_xlabel("time")
    axes[i].set_ylabel('queue_lenght')

plt.show()    
```

```python
fig , axes = plt.subplots( 4 ,4 ,   figsize = ( 20 , 12))
axes = axes.flatten()

cpu_utilization  = [
    col   for col in df.columns 
     if  "_utilization" in col.lower()
]
for i ,  col in enumerate(cpu_utilization) : 
    if i >= len(axes):
        continue
    axes[i].plot(pd.to_datetime(df['timestamp']), df[col], linewidth=1)
    axes[i].set_title( col)
    axes[i].set_xlabel("time")
    axes[i].set_ylabel('cpu_utilization')

plt.show()    
```
